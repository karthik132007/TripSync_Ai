```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
from db.get_from_db import get_places
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Dot
from tensorflow.keras.layers import Input

from tensorflow.keras.losses import MeanSquaredError
```

```python
places_df = get_places()
places_df=places_df.drop(columns=[0,1])
places=places_df.to_numpy()
```

```python
normalizer = tf.keras.layers.Normalization()
normalizer.adapt(places)
```

```python
weights = normalizer.get_weights()

np.savez(
    "norm_weights.npz",
    mean=weights[0],
    variance=weights[1],
    count=weights[2]
)
```

```python
model=Sequential([
    normalizer,
    Dense(128,activation='relu'),
    tf.keras.layers.Dropout(0.2),
    Dense(64)
])
```

```python
item_embeddings = model.predict(places)
```

```python
np.save("item_embeddings.npy", item_embeddings)
```
