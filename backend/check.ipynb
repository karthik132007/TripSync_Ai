```python
import psycopg2
import pandas as pd
from dotenv import load_dotenv
import os
load_dotenv()
conn = psycopg2.connect(
        host=os.getenv("DB_HOST"),
        database=os.getenv("DB_NAME"),
        user=os.getenv("DB_USER"),
        password=os.getenv("DB_PASS"),
        port=os.getenv("DB_PORT")
    )
cur = conn.cursor()
```

```python
try:
        rows=cur.execute("select id, place from places where lower(place) like '%goa%';")
        rows = cur.fetchone()
except Exception as e:
        print(e)
        conn.rollback()
```

```python
place_name = "cairo"
try:
    cur.execute(
        "select id, place from places where trim(lower(place)) = trim(lower(%s))",
        (place_name,),
    )
    row = cur.fetchone()
    place_id = row[0] if row else None
    print("exact:", row)
    if place_id is None:
        cur.execute(
            "select id, place from places where place ilike %s limit 5",
            (f"%{place_name}%",),
        )
        print("partial:", cur.fetchall())
except Exception as e:
    print(e)
    conn.rollback()
    place_id = None
print("place_id:", place_id)
```
